```python
import os
import pandas as pd
from quality_checks import preprocess_for_inference
from tensorflow.keras.applications import MobileNetV2
```

```python
model = MobileNetV2(weights="imagenet")
print("Model loaded successfully")
```

```python
DATA_DIR = "data"
image_files = os.listdir(DATA_DIR)
print(f"Total images found: {len(image_files)}")
```

```python
results = []

for img_name in image_files:
    img_path = os.path.join(DATA_DIR, img_name)
    accepted, metrics = preprocess_for_inference(img_path)

    row = {
        "image_name": img_name,
        "accepted_for_inference": accepted
    }

    if isinstance(metrics, dict):
        row.update(metrics)

    results.append(row)
```

```python
df = pd.DataFrame(results)
df.to_csv("prediction_results.csv", index=False)

df
```

### Observations

- The inference preprocessing pipeline demonstrates balanced behavior after threshold calibration.
- High-quality images pass all guardrails and are accepted for inference.
- Images with blur, extreme lighting, contrast issues, or excessive noise are rejected.
- Each rejection is explainable using individual quality metrics, ensuring transparent and safe inference.
